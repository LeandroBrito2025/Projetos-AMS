## As duas variantes e a latência

**Pareamento (acertos de cada variante, em quantos casos discordam e a favor de quem):** R: A acertou 47 e B acertou 50 dos 60 casos. Elas só discordam em 9 casos. B ganha em 6 (C11, C22, C32, C37, C42, C47): máquinas que vibravam um pouco acima de 7 e só precisavam de inspeção. A ganha em 3 (C15, C35, C49): máquinas que falharam de verdade, em que o certo era parar; mas mesmo nelas B mandou inspecionar, então nenhuma máquina foi liberada. Com só 9 discordâncias, um placar de 6 a 3 pode acontecer por acaso, como numa moeda jogada 9 vezes: a B parece melhor, mas ainda não dá para afirmar.

**Latência (p50 e p95, e por que a média engana em tempo de resposta):** R:p50 = 0,7 µs e p95 = 1,8 µs, em 3.000 decisões; média 0,8 µs e máximo 14,9 µs. A média engana em tempo de resposta porque poucos casos muito lentos a puxam para cima, como na fila do banco: uma pessoa que espera duas horas muda a média, mas não a espera de quase ninguém. O p50 diz o tempo típico, e o p95 diz o tempo que 95% das decisões não passam.
